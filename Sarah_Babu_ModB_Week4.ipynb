# 2. Weekly graph question

Below are a histogram and boxplot representation of the same data. A pharmacy is keeping a record of the prices of the drugs that it sells, and an administrator wants to know how much the more expensive drugs tend to cost, in the context of the other prices.

Please write a short explanation of the pros and cons of these two representations. Which would you choose? How would you modify the formatting, if at all, to make it more visually interesting, clear, or informative?

```python
import numpy as np
import pandas as pd

np.random.seed(0)
num_data = 100
data = np.exp(np.random.uniform(size = num_data) * 4)
df = pd.DataFrame(data.T, columns = ["data"])
```

```python
print("The 75th percentile is:", df.quantile(q = 0.75))
df.plot.hist()
```

Pro's of showing a histogram are: Shows distribution shape clearly: 1) You can see the strong right skew (many low-priced drugs, few very expensive ones).
2) Reveals clusters and gaps: Helpful if prices group into tiers.
3) Displays frequency: You see how common different price ranges are.

Cons are exact percentiles are hard to see (like the 75th percentile), sensitive to bin choice: Different bin widths can change interpretation, harder to quickly summarize central tendency and spread.

```python
df.plot.box()
```

Pros of showing box plot are: Excellent summary of spread (median, quartiles, IQR), highlights outliers clearly, makes it easy to see where the top 25% of drug prices begin (75th percentile).

Cons are does not show distribution shape,cannot see clustering or modality, less intuitive for non-technical audiences

# 4. Storytelling With Data graph

Reproduce any graph of your choice in p. 52-68 of the Storytelling With Data book as best you can.  (The second half of chapter two).  You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
categories = ["Category 1", "Category 2", "Category 3", "Category 4", "Category 5"]
A = [7, 5, 4, 4, 3]
B = [6, 6, 3, 5, 5]
C = [5, 3, 1, 1, 3]

y = np.arange(len(categories))
bar_height = 0.22

fig, ax = plt.subplots(figsize=(8, 5))

ax.barh(y - bar_height, A, height=bar_height, color="darkblue", label="A")
ax.barh(y, B, height=bar_height, color="lightblue", label="B")
ax.barh(y + bar_height, C, height=bar_height, color="lightgrey", label="C")

ax.set_yticks(y)
ax.set_yticklabels(categories)

ax.set_xlim(0, 8)
ax.set_xticks(range(0, 9))

ax.xaxis.tick_top()
ax.xaxis.set_label_position('top')

ax.invert_yaxis()

ax.set_title("Multiple Series Horizontal Bar Chart", pad=20)
ax.legend()

plt.tight_layout()
plt.show()
```
